```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('diabetes.csv')
df
```

```python
cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

for col in cols:
    df[col] = df[col].replace(0, df[col].median())
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
sns.countplot(x="Outcome", data=df)
plt.title("Diabetes Count (0 = No, 1 = Yes)")
plt.show()
```

```python
-This graph shows the number of people who have diabetes (1) and those who do not have diabetes (0) in the dataset.
```

```python
sns.histplot(df["Age"], bins=20,kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
-This graph shows the distribution of ages in the dataset, indicating how frequently different age groups appear.
```

```python
sns.histplot(df["Glucose"], bins=20, kde=True)
plt.title("Glucose Distribution")
plt.show()
```

```python
-This graph shows the distribution of glucose levels in the dataset, highlighting how frequently different glucose values occur.
```

```python
sns.boxplot(x="Outcome", y="BMI", data=df)
plt.tilte("BMI vs Diabetes")
plt.show()
```

```python
-This boxplot shows how BMI values differ between people with diabetes (1) and without diabetes (0)
```

```python
sns.boxplot(x="Outcome", y="Glucose", data=df)
plt.title("Glucose vs Diabetes")
plt.show()
```

```python
-This boxplot shows the comparison of glucose levels between people with diabetes (1) and those without diabetes (0).
```

```python
sns.boxplot(x="Outcome", y="Age", data=df)
plt.title("Age vs Diabetes")
plt.show()
```

```python
-This boxplot shows the comparison of age distribution between people with diabetes (1) and those without diabetes (0)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
-This heatmap shows the correlation between all numerical features in the dataset, indicating how strongly they are related to each other.
```
